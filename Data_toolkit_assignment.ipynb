**Data toolkit assingment**

1. A What is NumPy, and why is it widely used in Python?
> NumPy (Numerical Python) is a library for numerical operations, offering powerful N-dimensional arrays and vectorized operations, which are faster and more efficient than regular Python lists.

2. How does broadcasting work in NumPy?
> Broadcasting allows NumPy to perform arithmetic operations on arrays of different shapes by automatically expanding the smaller array to match the shape of the larger one.

3. What is a Pandas DataFrame?
> A DataFrame is a 2D labeled data structure with columns of potentially different types, similar to a spreadsheet or SQL table.

4. Explain the use of the groupby() method in Pandas.
> It splits data into groups based on some criteria, applies a function to each group, and then combines the results, useful for aggregation and summarization.

5. Why is Seaborn preferred for statistical visualizations?
> Seaborn provides high-level functions for attractive and informative statistical graphics with fewer lines of code.

6. What are the differences between NumPy arrays and Python lists?
> NumPy arrays are faster, more memory-efficient, and support vectorized operations, while Python lists are more flexible but slower for numerical computation.

7. What is a heatmap, and when should it be used?
> A heatmap is a graphical representation of data where values are represented as colors. It's useful for visualizing correlations or data density.

8. What does the term “vectorized operation” mean in NumPy?
> It refers to applying operations on entire arrays without using loops, leading to faster execution.

9. A How does Matplotlib differ from Plotly?
> Matplotlib creates static, publication-ready plots; Plotly offers interactive, web-based visualizations.

10. What is the significance of hierarchical indexing in Pandas?
> It allows multiple levels of indexing on an axis, enabling more complex data analysis and organization.

11. What is the role of Seaborn’s pairplot() function?
> It creates a grid of scatter plots for each pair of features, useful for exploring relationships in a dataset.

12. What is the purpose of the describe() function in Pandas?
> It generates summary statistics like mean, std, min, max, and quartiles for numeric columns.

13. Why is handling missing data important in Pandas?
> Missing data can lead to incorrect analysis; handling it ensures data integrity and reliable results.

14. What are the benefits of using Plotly for data visualization?
> Plotly provides interactive, customizable plots ideal for dashboards and web apps.

15.  How does NumPy handle multidimensional arrays?
> NumPy uses ndarray objects that can store elements in 1D, 2D, 3D, or higher dimensions, enabling complex numerical computations.

16. What is the role of Bokeh in data visualization?
> Bokeh is a Python library for creating interactive visualizations for web browsers.

17. Explain the difference between apply() and map() in Pandas?
> o	map() is used for element-wise operations on Series.
  
   o	apply() can be used on Series or DataFrame for row/column-wise custom functions.


18. What are some advanced features of NumPy?
> Broadcasting, memory-mapped files, linear algebra operations, FFT, masked arrays, and structured arrays.

19. How does Pandas simplify time series analysis?
> It supports date/time indexing, resampling, frequency conversion, and time-based indexing.

20. What is the role of a pivot table in Pandas?
> Pivot tables summarize data, allowing aggregation and reshaping of data using multiple indexes.

21. Why is NumPy’s array slicing faster than Python’s list slicing?
> NumPy slices views the same memory (no copying), whereas list slicing creates a new list.

22. What are some common use cases for Seaborn?
> Visualizing distributions, categorical data, regression plots, correlations, and heatmaps.
